```python
df_clean = df.copy()
```

```python
df_clean.isnull().sum()
```

```python
df_clean.drop_duplicates(inplace=True)
```

```python
df_clean.duplicated().sum()
```

```python
df_clean.drop(columns=["Surname"], inplace=True)
```

```python
df_clean.head()
```

```python
df_clean.info()
```

```python
df_clean["Age Group"] = pd.cut(
    df_clean["Age"],
    bins=[0,30,45,60,100],
    labels=[
        "Below 30",
        "30-45",
        "46-60",
        "Above 60"
    ]
)
```

```python
df_clean["Age Group"].value_counts()
```

```python
df_clean["Credit Score Band"] = pd.cut(
    df_clean["CreditScore"],
    bins=[0,500,700,900],
    labels=[
        "Low",
        "Medium",
        "High"
    ]
)
```

```python
df_clean["Credit Score Band"].value_counts()
```

```python
df_clean["Tenure Group"] = pd.cut(
    df_clean["Tenure"],
    bins=[-1,2,6,10],
    labels=[
        "New",
        "Mid-term",
        "Long-term"
    ]
)
```

```python
df_clean["Tenure Group"].value_counts()
```

```python
df_clean["Customer Value"] = np.where(
    df_clean["Balance"] >= 100000,
    "High Value",
    "Regular Value"
)
```

```python
df_clean["Customer Value"].value_counts()
```

```python
df_clean["Churn Status"] = df_clean["Exited"].map({
    0:"Retained",
    1:"Churned"
})
```

```python
df_clean["Churn Status"].value_counts()
```

```python
df_clean.columns
```

```python
df_clean.head()
```

```python
df_clean.to_csv(
    "cleaned_customer_churn.csv",
    index=False
)
```

```python
total_customers = len(df_clean)

print("Total Customers:", total_customers)
```

```python
churned_customers = df_clean["Exited"].sum()

print("Churned Customers:", churned_customers)
```

```python
retained_customers = total_customers - churned_customers

print("Retained Customers:", retained_customers)
```

```python
overall_churn_rate = (
    churned_customers / total_customers
) * 100

print(f"Overall Churn Rate: {overall_churn_rate:.2f}%")
```

```python
avg_credit_score = df_clean["CreditScore"].mean()

print(f"Average Credit Score: {avg_credit_score:.2f}")
```

```python
avg_balance = df_clean["Balance"].mean()

print(f"Average Balance: ${avg_balance:,.2f}")
```

```python
avg_salary = df_clean["EstimatedSalary"].mean()

print(f"Average Salary: ${avg_salary:,.2f}")
```

```python
active_member_rate = (
    df_clean["IsActiveMember"].mean()
) * 100

print(f"Active Member Rate: {active_member_rate:.2f}%")
```

```python
credit_card_rate = (
    df_clean["HasCrCard"].mean()
) * 100

print(f"Credit Card Ownership Rate: {credit_card_rate:.2f}%")
```

```python
avg_products = df_clean["NumOfProducts"].mean()

print(f"Average Products: {avg_products:.2f}")
```

```python
high_value = df_clean[
    df_clean["Balance"] >= 100000
]

high_value_churn = (
    high_value["Exited"].mean()
) * 100

print(f"High Value Customer Churn: {high_value_churn:.2f}%")
```

```python
geo_churn = (
    df_clean.groupby("Geography")["Exited"]
    .mean()
    * 100
).round(2)

geo_churn
```

```python
gender_churn = (
    df_clean.groupby("Gender")["Exited"]
    .mean()
    * 100
).round(2)

gender_churn
```

```python
age_churn = (
    df_clean.groupby("Age Group")["Exited"]
    .mean()
    * 100
).round(2)

age_churn
```

```python
credit_churn = (
    df_clean.groupby("Credit Score Band")["Exited"]
    .mean()
    * 100
).round(2)

credit_churn
```

```python
tenure_churn = (
    df_clean.groupby("Tenure Group")["Exited"]
    .mean()
    * 100
).round(2)

tenure_churn
```

```python
kpi_summary = pd.DataFrame({
    "KPI": [
        "Total Customers",
        "Churned Customers",
        "Retained Customers",
        "Overall Churn Rate",
        "Average Credit Score",
        "Average Balance",
        "Average Salary",
        "Active Member Rate",
        "Credit Card Ownership Rate",
        "Average Products",
        "High Value Customer Churn"
    ],
    "Value": [
        total_customers,
        churned_customers,
        retained_customers,
        f"{overall_churn_rate:.2f}%",
        f"{avg_credit_score:.2f}",
        f"${avg_balance:,.2f}",
        f"${avg_salary:,.2f}",
        f"{active_member_rate:.2f}%",
        f"{credit_card_rate:.2f}%",
        f"{avg_products:.2f}",
        f"{high_value_churn:.2f}%"
    ]
})

kpi_summary
```
